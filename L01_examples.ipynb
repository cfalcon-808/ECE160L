## 1. Comments

Python ignores text after `#` on a line. Triple-quoted text is technically a multiline string, although it is sometimes used like a long comment.

```python
# This entire line is a comment.
"""
This is a multiline string. Because it is not assigned to a variable,
it has no visible effect when this cell runs.
"""

'''
This is a multiline string. Because it is not assigned to a variable,
it has no visible effect when this cell runs.
'''

print("This line runs.")  # This comment follows Python code.

```

## 2. Printing output

The `print()` function displays values. Text inside quotation marks is a string.

```python
print("Hello, World!")
print('Single quotes work too.')
print("First line\nSecond line")  # \n starts a new line
```

## 3. Variables and assignment

The assignment operator `=` stores the value on the right in the variable on the left. Assigning a new value replaces the old one.

```python
message = "Hello, Python!"
print(message)

message = "Goodbye!"  # Replace the previous value.
print(message)
```

### Variable names versus string literals

`print(x)` displays the value stored in `x`, while `print("x")` displays the literal character **x**.

```python
x = "A random string"
print("x")
print(x)
```

## 4. User input

`input()` pauses the program and returns what the user enters as a string. A string argument to `input()` becomes the prompt.

```python
name = input("Enter your name: ")
print("Hello,", name)
```

Even if the user types digits, `input()` returns a string. Use `type()` to inspect a value's data type.

```python
favorite_number = int(input("Enter your favorite number: "))
print("You entered:", favorite_number)
print("Python stored it as:", type(favorite_number))
```

## 5. Printing multiple values

Separate values with commas inside `print()`. Python inserts a space between them by default.

```python
nickname = "CJ"
full_name = "Casey Jones"

print(nickname, full_name)
print("Hello", nickname + ".", "Is your full name", full_name + "?")
```

## 6. Combining strings

The `+` operator joins strings. Add `" "` when you want a space between names.

```python
first_name = input("Enter your first name: ")
last_name = input("Enter your last name: ")

full_name = first_name + " " + last_name
print("Hello, my name is", full_name)
```

### Another option: f-strings

An f-string inserts variable values inside `{}` and is often the clearest way to build a sentence.

```python
course = "ECE 160L"
topic = "Python basics"
print(f"Today in {course}, we are learning {topic}.")
```

## 7. A five-step plan for writing a program

Before writing code, use this simple plan:

1. **Define the goal:** State exactly what the program should do.
2. **Identify the input:** Decide what information the program needs.
3. **Plan the processing:** Decide how the input will be changed or combined.
4. **Produce the output:** Display the result clearly with `print()`.
5. **Test the program:** Try normal and unusual inputs and check the result.

### Basic example: a personalized greeting

**Goal:** Ask for a name and favorite hobby, then display one friendly sentence.

```python
# Step 1 - Goal: create a personalized greeting.

# Step 2 - Input: get the user's name and hobby.
name = input("What is your name? ")
hobby = input("What is your favorite hobby? ")

# Step 3 - Processing: combine the inputs into a sentence.
greeting = "Hello, " + name + "! It is great that you enjoy " + hobby + "."

# Step 4 - Output: show the completed sentence.
print(greeting)

# Step 5 - Test: run the cell with different names and hobbies.
# Example test: name = Maya, hobby = surfing
```

For the example test, the output should be: `Hello, Maya! It is great that you enjoy surfing.`
